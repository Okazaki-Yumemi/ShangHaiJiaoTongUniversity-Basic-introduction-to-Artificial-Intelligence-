## Gradio介绍

https://www.gradio.app/

   Gradio是一个开源的Python库，可以帮你快速创建一个简单漂亮的用户界面，展示你的机器学习模型。

   Gradio易于使用，只需简单定义输入和输出即可快速构建简单的交互页面。
   
   此外，Gradio还可以创建分享链接，直接分享给其他人使用。


</div>


**使用逻辑：**

输入组件(Input Component) → 处理函数(Function) → 输出组件(Output Component)
- 输入组件：用户交互的入口（文本框、上传文件、滑块等）
- 处理函数：自定义的 Python 函数，处理输入并返回结果
- 输出组件：展示处理结果的界面元素

![](https://ai-studio-static-online.cdn.bcebos.com/771012861a4e4f41ba6ce6eaa43629e522d62cf5d76746dea26ff7ac8979fee2)

**基础模块：**
* 应用界面：`gr.Interface`（搭建简易场景），`gr.Blocks`（制作定制化场景）。
* 输入输出：`gr.Image`（图像），`gr.Text`（文本框），`gr.Video`（视频框），`gr.DataFrame`（数据框），`gr.Dropdown`（下拉选项），`gr.Number`（数字），`gr.Markdown`，`gr.Files`，`gr.label`。
* 控制组件：`gr.Button`（按钮）。
* 布局组件：`gr.Tab`（标签页），`gr.Row`（行布局），`gr.Column`（列布局）等。

````python
import gradio as gr

def calculator(num1, operation, num2):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "multiply":
        return num1 * num2
    elif operation == "divide":
        if num2 == 0:
            raise gr.Error("Cannot divide by zero!")
        return num1 / num2

demo = gr.Interface(
    calculator,
    [
        "number",
        gr.Radio(["add", "subtract", "multiply", "divide"]), #创建单选按钮
        "number"
    ],
    "number",
    examples=[
        [45, "add", 3],
        [3.14, "divide", 2],
        [144, "multiply", 2.5],
        [0, "subtract", 1.2],
    ],
    title="Toy Calculator",
    description="Here's a sample toy calculator.",
    theme=gr.themes.Monochrome(),
)

demo.launch()
````

我们在这里使用上海交大提供的[DeepSeek API](https://claw.sjtu.edu.cn/guide/sjtu-api/)，获取 `base_url` 和 `api-key`

在 `api_key` 中填入实际申请到的 API Key，可以通过设置环境变量 `DEEPSEEK_API_KEY` 来实现安全导入。

可以在输入框中输入 `quit` 来结束对话并释放当前单元格。


|模型名称|模型参数|调用名|模式|侧重领域|上下文长度|
|---|---|---|---|---|---|
|DeepSeek V3.2（常规模式）|685B|deepseek-chat|非思考模式|通用文本处理|32k|
|DeepSeek V3.2（思考模式）|685B|deepseek-reasoner|深度思考模式|复杂逻辑深度推理|32k|

````python
from openai import OpenAI

def main():   
    base_url = "https://models.sjtu.edu.cn/api/v1"
    api_key = 'your-api-key'             # 这里替换成你自己申请的Key，或通过环境变量安全导入。
    model_id = "deepseek-chat"              # 这里替换成你想使用的模型ID，建议先在平台上测试不同模型的表现，再决定使用哪个模型进行开发
    
    # ========== 可调参数（试试不同值，观察模型输出的差异）==========
    temperature = 0.1   # 范围 0~2，值越高输出越随机/有创意，值越低输出越确定/保守
    top_p = 0.5          # 范围 0~1，核采样阈值，值越低只从概率最高的词中选取，输出更集中
    # =============================================================
    
    try:
        client = OpenAI(
            api_key=api_key,
            base_url=base_url
        )
    except Exception as e:
        print(f"Error: {e}")
        return
    
    messages = []
    print("\nStart chatting ('quit' to exit)")
    print("-" * 40)
    
    while True:
        try:
            user_input = input("\nYou: ").strip()
            
            if user_input.lower() == 'quit':
                print("Goodbye!")
                break
            
            if not user_input:
                continue
            
            messages.append({"role": "system", "content": "You are a helpful assistant"})   # 系统提示词
            messages.append({"role": "user", "content": user_input})                        # 用户提示词
            
            print("Assistant: ", end="", flush=True)
            
            # 开启 stream=True
            stream = client.chat.completions.create(
                model=model_id,
                messages=messages,
                stream=True,
                temperature=temperature,
                top_p=top_p,
            )
            # 逐步打印输出
            full_response = ""
            for chunk in stream:
                if chunk.choices[0].delta.content:
                    content = chunk.choices[0].delta.content
                    print(content, end="", flush=True)
                    full_response += content
            
            print()
            # 拼接上下文，将完整回复添加到消息列表中，供后续对话使用
            if full_response:
                messages.append({"role": "assistant", "content": full_response})
                
        except KeyboardInterrupt:
            print("\n\nGoodbye...")
            break
        except Exception as e:
            print(f"\nError: {e}")

if __name__ == "__main__":
    main()
````

给Chatbot装上可视化界面

在命令行里聊天相对枯燥，Gradio的 `ChatInterface` 组件支持并简化了对接大模型流式输出的工作。

</div>



````python
import gradio as gr
from openai import OpenAI

# 1. 配置模型参数
API_KEY = 'your-api-key'           # 请注意替换为自己的key
BASE_URL = "https://models.sjtu.edu.cn/api/v1"
MODEL_ID = "deepseek-chat"                        # "deepseek-chat", "deepseek-reasoner", "glm", "qwen", "minimax"

client = OpenAI(api_key=API_KEY, base_url=BASE_URL)

def chat_stream(message, history):
    """
    处理函数：
    message: 当前用户的输入文本
    history: 历史对话列表
    """
    # 将 Gradio 提供的前端历史记录转换为 OpenAI API 支持的 messages 格式
    messages =[{"role": "system", "content": "你是一个聪明、幽默的AI助手。"}]
    for user_msg, bot_msg in history:
        messages.append({"role": "user", "content": user_msg})
        messages.append({"role": "assistant", "content": bot_msg})
        
    # 将当前的新问题追加到最后
    messages.append({"role": "user", "content": message})

    # 发起流式请求
    stream = client.chat.completions.create(
        model=MODEL_ID,
        messages=messages,
        stream=True
    )

    partial_message = ""
    for chunk in stream:
        if chunk.choices[0].delta.content is not None:
            partial_message += chunk.choices[0].delta.content            
            yield partial_message # yield 关键字会将得到的字符串片段逐步推送到前端，实现打字机效果

# 2. 使用 ChatInterface 创建对话界面
demo = gr.ChatInterface(
    fn=chat_stream,
    title="DeepSeek 网页端助手",
    description="基于 Gradio 和 DeepSeek API 搭建的流式对话系统。",
    examples=["你好！", "请给我讲个笑话", "介绍一下上海交通大学"]
)

if __name__ == "__main__":
    # 在 Jupyter Notebook 中会以内嵌网页的形式展示
    demo.launch()
````

### 这个综合项目到底在做什么

你可以把它理解为一个两阶段应用：

1. 第一阶段：用户上传图片，YOLO 先识别图片中有哪些目标。
2. 第二阶段：把检测结果整理成文字，再交给 LLM 生成自然语言描述、分析或问答结果。

### 建议的模块拆分

1. 模型准备模块：加载 YOLO 模型，初始化 LLM 客户端。

def summarize_detections(result):
    """提取 YOLO 检测结果中的类别、数量、置信度和位置信息。"""

def sanitize_model_output(text):
    """清洗思考型模型可能返回的思维过程、标题和代码块。"""
    if not text:
        return ""

    clean_text = str(text).strip()
    clean_text = clean_text.replace("&lt;think&gt;", "<think>").replace("&lt;/think&gt;", "</think>")
    clean_text = re.sub(r"(?is)<think>.*?(?:</think>|$)", "", clean_text)
    clean_text = re.sub(r"```(?:markdown|md|text)?", "", clean_text, flags=re.IGNORECASE)
    clean_text = clean_text.replace("```", "")

    section_patterns = [
        r"(?is)^.*?(?:最终答案[:：]|最终描述[:：]|输出[:：])",
        r"(?is)^思考过程[:：].*$",
        r"(?is)^推理过程[:：].*$",
        r"(?is)^分析过程[:：].*$",
        r"(?is)^reasoning[:：].*$",
        r"(?is)^thought[:：].*$",
    ]
    for pattern in section_patterns:
        clean_text = re.sub(pattern, "", clean_text)

    lines = []
    for line in clean_text.splitlines():
        stripped = line.strip()
        lower = stripped.lower()
        if not stripped:
            continue
        if lower.startswith(("思考过程", "推理过程", "分析过程", "reasoning", "thought", "reflection")):

            continue
        lines.append(stripped)

    clean_text = " ".join(lines).strip()
    clean_text = re.sub(r"\s+", " ", clean_text)
    return clean_text.strip("-:： ")

    choice = response.choices[0]

    content = getattr(message, "content", "") or ""
    raw_content = content

    if isinstance(content, list):
        text_fragments = []
        for part in content:
            if isinstance(part, dict) and part.get("type") == "text":
                text_fragments.append(part.get("text", ""))
            elif hasattr(part, "text") and getattr(part, "text", None):
                text_fragments.append(part.text)
            else:
                text_fragments.append(str(part))
        content = "".join(text_fragments)
        raw_content = content

    cleaned_content = sanitize_model_output(content)
    if cleaned_content:
        return cleaned_content

    reasoning = getattr(message, "reasoning_content", None)

def compact_description(text, max_chars=96):
    """压缩展示文本，避免在输出框中出现过长文本和截断感。"""
    clean_text = " ".join((text or "").split()).strip()
    if not clean_text or clean_text.startswith("[") or len(clean_text) <= max_chars:
        return clean_text

    candidate = clean_text[:max_chars]
    last_stop = max(candidate.rfind(mark) for mark in "。！？；")
    last_pause = max(candidate.rfind(mark) for mark in "，、：")

    if last_stop >= max_chars // 2:
        candidate = candidate[: last_stop + 1]
    elif last_pause >= max_chars // 2:
        candidate = candidate[: last_pause + 1]
    else:
        candidate = candidate.rstrip() + "……"

    return candidate.strip()

    except Exception as error:
        if not detection_text or detection_text == build_detection_text({}, []):
            return f"[文本模型 {model_name} 调用失败] 图片中似乎没有发现明显物体。错误信息：{error}"
        return f"[文本模型 {model_name} 调用失败] 可先根据下方检测明细手动分析。错误信息：{error}"

def get_vision_llm_description(img, custom_prompt, model_name):
    """直接把图像送给多模态模型生成描述 B。"""
    image_base64 = image_to_base64(img)
    text_prompt = f"""
你是描述生成角色 B。请直接观察图片并生成一段中文描述。
要求：
1. 重点说明主要物体、数量、相对位置和可直接观察到的外观细节。
2. 只输出最终描述，不要输出比较、结论、思考过程或分析框架。
3. 尽量控制在 80 到 100 字。

补充要求：{format_generation_instruction(custom_prompt)}
"""

    try:
        response = client.chat.completions.create(
            model=model_name,
            messages=[
                {
                    "role": "user",
                    "content": [
                        {"type": "text", "text": text_prompt},
                        {
                            "type": "image_url",
                            "image_url": {
                                "url": f"data:image/png;base64,{image_base64}"
                            },
                        },
                    ],
                }
            ],
            max_tokens=140,
            temperature=0.45,
        )
        return extract_response_text(response, task_kind="vision")
    except Exception as error:
        return f"[多模态模型 {model_name} 调用失败] 无法直接生成图片描述。错误信息：{error}"

def get_comparison_summary(detection_overview, desc_a, desc_b, custom_prompt, text_model_name):
    """在两段描述都生成后，再单独调用 LLM 生成 A/B 比较分析。"""
    comparison_rule = (
        "- 只输出最终 Markdown 列表。若输出中夹带 <think></think> 或内部推理片段，系统会在展示前自动清洗。"
        if uses_post_cleaning(text_model_name)
        else "- 只输出最终 Markdown 列表，不要输出思考过程、分析前言、标题或代码块。"
    )

    prompt = f"""
你是课堂点评角色。下面有一份简短检测摘要，以及两种不同路径生成的描述：

检测摘要：
{detection_overview}

描述 A（只基于 YOLO 结果）：
{desc_a}

描述 B（多模态直接看图）：
{desc_b}

补充要求：{format_generation_instruction(custom_prompt)}

请输出 4 个 Markdown 列表项，每项 1 到 2 句，要求有分析深度，不要空泛套话：
- 结构可靠性：判断谁最依赖可核对的证据，谁最容易受检测误差影响。
- 细节感知力：判断谁最能补充视觉细节，并指出这种补充可能带来的推断风险。
- 应用场景：评价 A 和 B 各自更适合什么课堂场景或任务目标。
- 教学启发：给学生一个有操作性的比较方法，告诉他们先看什么、再看什么。

限制：
{comparison_rule}
- 不要复读原文，不要简单写“A最好/B最好”。
- 要点要具体，尽量结合这张图里出现的对象、位置和细节差异。
- 总长度控制在 220 字以内。
"""

    try:
        response = client.chat.completions.create(
            model=text_model_name,
            messages=[
                {"role": "system", "content": get_comparison_system_message(text_model_name)},
                {"role": "user", "content": prompt},
            ],
            max_tokens=get_comparison_max_tokens(text_model_name),
            temperature=0.45,
        )
        return extract_response_text(response, task_kind="comparison")
    except Exception:
        return (
            "- **结构可靠性**：A 最便于核对，因为它紧贴 YOLO 的类别和位置；但如果检测漏掉目标，A 也会同步漏写。\n"
            "- **细节感知力**：B 最能补充外观与材质，例如小狗衣着和背景质感；代价是它更可能加入检测结果里没有的主观推断。\n"
            "- **应用场景**：A 更适合做结构化识别讲解，B 更适合做图像观察与语言表达训练。\n"
            "- **教学启发**：先核对 YOLO 明细里的类别与位置，再比较两段描述谁更准确、谁更丰富。"
        )

def process_image(img, custom_prompt, text_model_name, vision_model_name):
    if img is None:
        message = "请先上传图片。"
        return None, "{}", message, message, message

    try:
        results = model.predict(img, device=DEVICE, verbose=False)
        annotated_img = results[0].plot()

        counts, detections = summarize_detections(results[0])
        detection_text = build_detection_text(counts, detections)
        detection_overview = build_detection_overview(counts, detections)

        text_description_raw = get_text_llm_description(
            detection_text=detection_text,
            custom_prompt=custom_prompt,
            model_name=text_model_name,
        )
        vision_description_raw = get_vision_llm_description(
            img=img,
            custom_prompt=custom_prompt,
            model_name=vision_model_name,
        )

        text_description = compact_description(text_description_raw, max_chars=88)
        vision_description = compact_description(vision_description_raw, max_chars=96)

        comparison_summary = get_comparison_summary(
            detection_overview=detection_overview,
            desc_a=text_description,
            desc_b=vision_description,
            custom_prompt=custom_prompt,
            text_model_name=text_model_name,
        )

        return (
            annotated_img,
            detection_text,
            text_description,
            vision_description,
            comparison_summary,
        )
    except Exception as error:
        message = f"处理失败：{error}"
        return None, "{}", message, message, message
